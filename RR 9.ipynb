## Friends 

*Friends* is an American sitcom that aired from 1994 to 2004





```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("C:\\Users\\Ceee\\Desktop\\sem 4\\Reproducible Research\\FRIENDS.xlsx")

print(df.head(15))   # shows first 5 rows
```

```python


# select only what you need
df_clean = df[["season", "episodes", "rating", "viewers (millions)"]].copy()

# rename columns (optional but recommended)
df_clean.columns = ["Season", "Episodes", "Rating", "Viewers"]

# convert to numeric (important for plotting)
df_clean["Season"] = pd.to_numeric(df_clean["Season"], errors="coerce")
df_clean["Episodes"] = pd.to_numeric(df_clean["Episodes"], errors="coerce")
df_clean["Rating"] = pd.to_numeric(df_clean["Rating"], errors="coerce")
df_clean["Viewers"] = pd.to_numeric(df_clean["Viewers"], errors="coerce")

# drop any broken rows
df_clean = df_clean.dropna()

df_clean
```

```python


plt.plot(df_clean["Season"], df_clean["Viewers"])
plt.xlabel("Season")
plt.ylabel("Viewers (millions)")
plt.title("Friends Viewership Over Time")
plt.show()
```

```python
#changes
df_clean["Change"] = df_clean["Viewers"].diff()

plt.bar(df_clean["Season"], df_clean["Change"])
plt.xlabel("Season")
plt.ylabel("Change in viewers")
plt.title("Season-to-Season Change in Viewership")
plt.show()
```

The viewership of *Friends* increased from approximately 24.8 million in season 1 
to a peak of 31.7 million in season 2.

After the peak, the audience gradually declined, reaching around 22.1 million 
in season 7. This represents a decrease of about 9.6 million viewers.

In later seasons, the show regained popularity, with viewership rising again 
to over 26 million by the final season.
